```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Step 1: Define the parameters of the normal distribution
mean = 15.0      # Average battery life in hours
std_dev = 2.0   # Standard deviation in hours
num_phones = 1000

# Step 2: Generate 1000 random battery life measurements
data = np.random.normal(mean, std_dev, num_phones)

# Step 3: Display the first 10 generated battery lives
print("First 10 generated battery life values (hours):")
print(np.round(data[:10], 2))

# Step 4: Display the parameters
print("\nParameters of the Normal Distribution")
print("-------------------------------------")
print("Mean:", mean)import numpy as np

# Given values
P_spam = 0.20
P_offer_spam = 0.80
P_offer_not_spam = 0.10

# Probability of getting "offer"
P_offer = (P_offer_spam * P_spam) + (P_offer_not_spam * (1 - P_spam))

# Bayes theorem
P_spam_offer = (P_offer_spam * P_spam) / P_offer

print("P(Spam | offer) =", P_spam_offer * 100, "%")
print("Standard deviation:", std_dev)
print("Number of phones:", num_phones)

# Step 5: Calculate basic statistics of generated data
sample_mean = np.mean(data)
sample_std_dev = np.std(data)
minimum_life = np.min(data)
maximum_life = np.max(data)

# Step 6: Display the calculated statistics
print("\nStatistics of Generated Battery Life")
print("-------------------------------------")
print("Sample mean:", round(sample_mean, 2))
print("Sample standard deviation:", round(sample_std_dev, 2))
print("Minimum battery life:", round(minimum_life, 2))
print("Maximum battery life:", round(maximum_life, 2))

# Step 7: Create a figure for plotting
plt.figure(figsize=(10, 6))

# Step 8: Plot the histogram of generated battery lives
plt.hist(
    data,
    bins=30,
    density=True,
    alpha=0.6,
    edgecolor="black",
    label="Generated battery data"
)

# Step 9: Generate values for the theoretical normal curve
x = np.linspace(8, 22, 200)

# Step 10: Calculate probability density values
y = norm.pdf(x, mean, std_dev)

# Step 11: Plot the theoretical normal distribution curve
plt.plot(
    x,
    y,
    linewidth=2,
    label="Normal curve"
)

# Step 12: Mark the mean on the graph
plt.axvline(
    mean,
    color="red",
    linestyle="--",
    linewidth=2,
    label="Mean = 15.0 hrs"
)

# Step 13: Add labels to the horizontal and vertical axes
plt.xlabel("Battery Life (Hours)")
plt.ylabel("Probability density")

# Step 14: Add a title to the graph
plt.title("Normal Distribution of Smartphone Battery Life")

# Step 15: Display the legend
plt.legend()

# Step 16: Add a light grid for better readability
plt.grid(alpha=0.2)

# Step 17: Adjust the graph layout
plt.tight_layout()

# Step 18: Display the completed graph
plt.show()

# Step 19: Select a specific battery life value for z-score calculation (e.g., a defective phone lasting 10 hours)
battery_life = 10.0

# Step 20: Calculate the z-score
z_score = (battery_life - mean) / std_dev

# Step 21: Display the z-score calculation
print("\nZ-Score Calculation")
print("-------------------")
print("Observed battery life:", battery_life, "hours")
print("Mean:", mean)
print("Standard deviation:", std_dev)
print("Z-score for 10.0 hours:", round(z_score, 2))

# Step 22: Interpret the z-score
if z_score > 0:
    print("The battery life is above the mean.")
elif z_score < 0:
    print("The battery life is below the mean (potential hardware defect).")
else:
    print("The battery life is equal to the mean.")

# Step 23: Display the final conclusion
print("\nConclusion:")
print("The histogram represents the generated smartphone battery life samples.")
print("The curve represents the theoretical normal distribution.")
print("The z-score shows the distance from the mean in standard deviations (e.g., Z = -2.5 indicates a severe anomaly).")
```
